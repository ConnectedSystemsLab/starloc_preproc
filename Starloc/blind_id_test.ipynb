```python
from datetime import datetime
import pytz
d_bound_1 = datetime(2024,3,13,0,0,0,tzinfo=pytz.utc)
time_stamp_start = d_bound_1.timestamp()
d_bound_2 = datetime(2024,3,14,0,0,0,tzinfo=pytz.utc)
time_stamp_end = d_bound_2.timestamp()
tle_path = '13Mar2024_06_14/Starlink.tle'
fc = 11.325e9
```

```python
import random
from sat_id import faster_doppler_sim
from sat_id import aoa_sim_1_sat
import numpy as np
from matplotlib import pyplot as plt
duration = 15*60 # 15 mins
captured_sat_num = 6
lat_gt, lon_gt = 40, -88
num_exp = 10
res = np.zeros(num_exp)
for exp_idx in range(num_exp):

    sim_start_time = random.uniform(time_stamp_start, time_stamp_end)
    start_time_datetime = datetime.fromtimestamp(sim_start_time, tz=pytz.utc)
    # simulate all satellite's passing
    sat_dopp_dict, t_list = faster_doppler_sim(1, tle_path, fc, sim_start_time, sim_start_time+duration, 10, ground_station=(lat_gt, lon_gt), min_ele=75)
    # get the key of sat_dopp_dict
    sat_id_list = list(sat_dopp_dict.keys())
    # randomly select captured_sat_num satellites
    captured_sat_id_list = random.sample(sat_id_list, captured_sat_num)
    # get a more accurate time for the simulation
    az_gt_list, el_gt_list, range_gt_list, time_arr, t_sim_list = [], [], [], [], []
    is_middle_arr = np.array([])
    for captured_sat_id in captured_sat_id_list:
        t_sat_start = t_list[np.where(np.logical_not(np.isnan(sat_dopp_dict[captured_sat_id][0])))[0][0]]
        t_sat_end = t_list[np.where(np.logical_not(np.isnan(sat_dopp_dict[captured_sat_id][0])))[0][-1]]
        t_sim, az_gt, el_gt, range_gt = aoa_sim_1_sat(1, tle_path, captured_sat_id, sim_start_time+t_sat_start, sim_start_time+t_sat_end, .1, lat_gt, lon_gt)
        t_sim += t_sat_start
        az_gt, el_gt = np.deg2rad(az_gt), np.deg2rad(el_gt)
        az_gt_list.append(az_gt)
        el_gt_list.append(el_gt)
        range_gt_list.append(range_gt)
        time_arr.append(t_sim[np.argmax(el_gt)])
        t_sim_list.append(t_sim)
        is_middle_arr = np.append(is_middle_arr, np.argmax(el_gt) != 0 and np.argmax(el_gt) != len(el_gt)-1)

    time_arr = np.array(time_arr)
    from utils.satid_blind_identification import get_candidate_locs, get_candidate_locs_jida
    search_radius = 250 # in km
    gs_groups, coarse_gs, sat_groups = get_candidate_locs_jida(start_time_datetime, time_arr, is_middle_arr, tle_path, search_dist_bound=search_radius)

    # check whether there's a correct group
    group_num = len(sat_groups)
    correct_group_idx = None
    for i in range(group_num):
        sat_groups_i = sat_groups[i]
        sat_groups_name_i = [sat_groups_i[j].name for j in range(captured_sat_num)]
        if set(captured_sat_id_list) == set(sat_groups_name_i):
            print('Correct group found!')
            correct_group_idx = i
            res[exp_idx] = 1
            break
```
